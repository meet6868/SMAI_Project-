This notebook demonstrates PDF ingestion, vector indexing, retrieval, and grounded answer generation with citations.

```python
from pathlib import Path
import sys

project_root = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.voter_assistant.knowledge_assistant import RAGPipeline
```

```python
pipeline = RAGPipeline(data_dir=project_root / 'data', db_dir=project_root / 'vector_store')
pipeline.build_index(force_rebuild=False)
print('Index ready')
```

```python
query = 'How can I apply for a new voter ID?'
contexts = pipeline.retrieve(query=query, top_k=4)
for i, c in enumerate(contexts, start=1):
    print(f"{i}. {c['source']} page {c['page']} score={c['score']:.3f}")
    print(c['text'][:300] + '...')
    print()
```
